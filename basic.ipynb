```python
import numpy as np
```

```python
image = np.array([
    [2, 3, 2, 4, 3],
    [3, 5, 8, 5, 2],
    [2, 8, 20, 9, 3],
    [3, 5, 9, 6, 2],
    [2, 3, 2, 3, 2]
])
```

```python
print("Image dimensions:", image.shape)
```

```python
print("Brightest pixel:", image.max())
```

```python
print("Average value:", image.mean())
```

```python
import matplotlib.pyplot as plt
```

```python
plt.imshow(image, cmap="inferno", origin="lower")
plt.colorbar(label="Illustrative pixel value")
plt.title("A tiny example of an astronomical image")
plt.xlabel("Pixel X")
plt.ylabel("Pixel Y")
plt.show()
```
